## 4. Impact footprint (Sentinel-2 and Sentinel-1)

- **Optical.** Sentinel-2 medians for the same season one year apart (27 Aug–28 Oct, 2025 against 2026), clouds removed with Cloud Score+. A pixel is flagged where vegetation was lost *or* its whole spectrum changed strongly (change vector analysis, which catches debris on ground that was already bare), and the surface is now bare, sediment or water.
- **Radar.** Sentinel-1 VH change on a single orbit, July–August against 26 Aug–15 Sep. It adds little here: steep gorge walls cause layover and shadow.
- **Where.** Only the event corridor (the Lhende Khola and the Bhote Koshi–Trishuli main stem, ±1 km) and only the valley floor (height above nearest drainage below 150 m), so other rivers and the late-September rains are not counted.
- **Thresholds** were chosen by `python -m rasuwa tune` against Copernicus EMS EMSR927 with leave-one-area-out, and a limit on false alarms in the untouched Langtang valley. Scores: `outputs/validation.json`.

Areas come from `python -m rasuwa export` then `python -m rasuwa footprint`. The footprint itself is in `data/footprint/` as GeoTIFF tiles that open in QGIS.
